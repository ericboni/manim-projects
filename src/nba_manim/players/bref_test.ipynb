```python
import pandas as pd
import io
import requests
import lxml
```

```python
BASE_URL = "https://www.basketball-reference.com/"

url = f'{BASE_URL}/players/a/'

params = None
headers = {"User-Agent": "Mozilla/5.0 (personal research script)"}
resp = requests.get(url, params=params, headers=headers, timeout=15)
resp.raise_for_status()
```

```python
tables = pd.read_html(io.StringIO(resp.text),extract_links="a", flavor="lxml")
for table in tables:
    print(table.tail()) 
```
