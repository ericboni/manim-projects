```python
import pandas as pd
import io
import requests
import lxml
import time
```

```python
BASE_URL = "https://www.basketball-reference.com/"

url = f'{BASE_URL}/players/a/'

params = None
headers = {"User-Agent": "Mozilla/5.0 (personal research script)"}
resp = requests.get(url, params=params, headers=headers, timeout=15)
resp.raise_for_status()
```

```python
resp
```

```python
df_players = pd.DataFrame()
for letter in "abcdefghijklmnopqrstuvwxyz":
    print(f"Fetching players for letter: {letter}")
    url = f'{BASE_URL}/players/{letter}/'
    resp = requests.get(url, params=params, headers=headers, timeout=15)
    resp.raise_for_status()
    tables = pd.read_html(io.StringIO(resp.text), extract_links="body", flavor="lxml")
    for table in tables:
        for col in table.columns:
            if col == 'Player':
                table['bref_id'] = table[col].apply(lambda x: x[1].split('/')[-1].split('.')[0])  # Extract the id from the player
                table[col] = table[col].apply(lambda x: x[0]) #Player name

            else:
                table[col] = table[col].apply(lambda x: x[0])

        df_players = pd.concat([df_players, table], ignore_index=True)
    time.sleep(8)  # Be nice to the server and avoid rate limiting


```

```python
df_players.to_csv("registry.csv",index=False)
```
